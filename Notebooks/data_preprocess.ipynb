```python
import os
import pandas as pd
import numpy as np

# paths
RAW_PATH = "../data/application_train.csv"
BATCH_DIR = "../data/batches"

os.makedirs(BATCH_DIR, exist_ok=True)

# load data
df = pd.read_csv(RAW_PATH)

# shuffle to simulate real-world arrival
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

# batch size (recommended)
BATCH_SIZE = 25000

def create_batches(df, batch_size):
    num_batches = int(np.ceil(len(df) / batch_size))
    
    for i in range(num_batches):
        start = i * batch_size
        end = (i + 1) * batch_size
        
        batch = df.iloc[start:end].copy()
        
        # simulate ingestion time (important)
        batch["ingestion_time"] = pd.Timestamp("2024-01-01") + pd.Timedelta(days=i)
        
        # save batch
        batch_path = os.path.join(BATCH_DIR, f"batch_{i}.csv")
        batch.to_csv(batch_path, index=False)
        
        print(f"Saved batch_{i} with shape {batch.shape}")

create_batches(df, BATCH_SIZE)
```
